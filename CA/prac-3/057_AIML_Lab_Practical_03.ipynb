# **Question 1: Linear Regression**

## Problem Statement

The objective is to predict a student's **Course Reward** based on marks obtained in four tests:

- T1
- T2
- T3
- T4

A dataset containing 10 samples will be created.

Linear Regression will be used to learn the relationship between the four test marks and the Course Reward.

The trained model will then be used to predict the Course Reward for new input marks.

## Aim

To implement Multiple Linear Regression for predicting Course Reward using marks obtained in T1, T2, T3 and T4.

## Theory

Linear Regression is a supervised machine learning algorithm used to predict a continuous numerical output.

Since we have four input variables (T1, T2, T3 and T4), this problem uses **Multiple Linear Regression**.

The general equation is:

### y = b0 + b1T1 + b2T2 + b3T3 + b4T4

Where:

- y = predicted Course Reward
- b0 = intercept
- b1, b2, b3, b4 = coefficients
- T1, T2, T3, T4 = input marks

The model learns the coefficients from the training data.

After training, the model can predict the Course Reward for new marks.

```python
# Create 10 Samples
import pandas as pd

# Dataset containing 10 samples

data = {
    "T1": [65, 72, 80, 55, 90, 68, 75, 85, 60, 95],
    "T2": [70, 75, 82, 60, 88, 72, 78, 90, 65, 92],
    "T3": [68, 78, 85, 58, 92, 70, 80, 88, 63, 94],
    "T4": [72, 74, 84, 62, 90, 75, 82, 91, 67, 96],
    "Course_Reward": [68, 75, 83, 59, 90, 72, 79, 89, 64, 94]
}

df = pd.DataFrame(data)

print("Dataset:")
display(df)
```

```python
# Separate Input and Output
# Input variables
X = df[["T1", "T2", "T3", "T4"]]

# Output variable
y = df["Course_Reward"]

print("Input Variables:")
display(X)

print("\nOutput Variable:")
display(y)
```

```python
# Train Linear Regression
from sklearn.linear_model import LinearRegression

# Create Linear Regression model
model = LinearRegression()

# Train the model
model.fit(X, y)

print("Linear Regression model trained successfully.")

print("\nIntercept:")
print(model.intercept_)

print("\nCoefficients:")
print(model.coef_)
```

```python
# Prediction for Any Input Marks
# Enter new marks
T1 = 80
T2 = 85
T3 = 82
T4 = 88

# Create input for prediction
new_input = pd.DataFrame({
    "T1": [T1],
    "T2": [T2],
    "T3": [T3],
    "T4": [T4]
})

# Predict Course Reward
prediction = model.predict(new_input)

print("Input Marks:")
print("T1 =", T1)
print("T2 =", T2)
print("T3 =", T3)
print("T4 =", T4)

print("\nPredicted Course Reward:")
print(round(prediction[0], 2))
```

```python
# Actual vs Predicted
# Predict rewards for the original 10 samples

df["Predicted_Reward"] = model.predict(X)

print("Actual vs Predicted Course Reward:")
display(df)
```

```python
# Graph
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.plot(
    range(1, 11),
    df["Course_Reward"],
    marker="o",
    label="Actual Reward"
)

plt.plot(
    range(1, 11),
    df["Predicted_Reward"],
    marker="x",
    label="Predicted Reward"
)

plt.xlabel("Sample Number")
plt.ylabel("Course Reward")

plt.title("Actual vs Predicted Course Reward")

plt.legend()
plt.grid(True)

plt.show()
```

## Result

Multiple Linear Regression was successfully implemented using T1, T2, T3 and T4 marks as input variables.

A dataset containing 10 samples was used to train the model.

The trained model successfully predicted the Course Reward for new input marks.

Therefore, Linear Regression can be used to predict Course Reward based on the student's test marks.

# **Question 2 : Non-Linear Regression**

## Dataset: Concrete Compressive Strength

The Concrete Compressive Strength dataset from the UCI Machine Learning Repository is used for this experiment.

The dataset contains information about concrete ingredients and curing age.

### Link: https://www.kaggle.com/datasets/niteshyadav3103/concrete-compressive-strength?resource=download

### Input Variables

1. Cement
2. Blast Furnace Slag
3. Fly Ash
4. Water
5. Superplasticizer
6. Coarse Aggregate
7. Fine Aggregate
8. Age

### Output Variable

**Concrete Compressive Strength (MPa)**

The dataset contains 1030 samples.

The relationship between the ingredients, age and concrete compressive strength is highly nonlinear.

## Aim

To implement Non-Linear Regression using the Concrete Compressive Strength dataset and visualize the relationship between the input variable and predicted output.

## Theory: Non-Linear Regression

Non-Linear Regression is used when the relationship between the input and output variables cannot be represented adequately by a straight line.

In this experiment, Polynomial Regression is used.

Polynomial Regression extends Linear Regression by adding powers of the input variable.

For example:

y = b0 + b1x + b2x² + b3x³

Here:

- x = input variable
- y = predicted output
- b0, b1, b2, b3 = model coefficients

The polynomial curve can represent nonlinear relationships between variables.

For visualization, we will use the **Age** of concrete as the input variable and **Concrete Compressive Strength** as the output variable.

```python
!pip install -q ucimlrepo
```

```python
from ucimlrepo import fetch_ucirepo

# Fetch Concrete Compressive Strength dataset
concrete = fetch_ucirepo(id=165)

# Features
X = concrete.data.features

# Target
y = concrete.data.targets

print("Dataset Shape:")
print(X.shape)

print("\nFeatures:")
display(X.head())

print("\nTarget:")
display(y.head())
```

```python
# Select Age and Strength
import pandas as pd

# Select Age as input
age = X[["Age"]]

# Select concrete strength as output
strength = y.iloc[:, 0]

print("Age:")
display(age.head())

print("\nConcrete Compressive Strength:")
display(strength.head())
```

```python
#Polynomial Regression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import make_pipeline

# Create Polynomial Regression model
degree = 3

model_poly = make_pipeline(
    PolynomialFeatures(degree),
    LinearRegression()
)

# Train model
model_poly.fit(age, strength)

print("Polynomial Regression model trained successfully.")
```

```python
# Generate predictions
predicted_strength = model_poly.predict(age)

print("First 10 predictions:")

for i in range(10):
    print(
        "Actual:",
        round(strength.iloc[i], 2),
        "Predicted:",
        round(predicted_strength[i], 2)
    )
```

```python
# Non-Linear Regression Graph
import numpy as np
import matplotlib.pyplot as plt

# Create smooth age values
age_range = np.linspace(
    age["Age"].min(),
    age["Age"].max(),
    300
).reshape(-1, 1)

# Predict strength for smooth values
strength_curve = model_poly.predict(age_range)

# Scatter plot of actual data
plt.figure(figsize=(9, 6))

plt.scatter(
    age["Age"],
    strength,
    alpha=0.5,
    label="Actual Data"
)

# Polynomial regression curve
plt.plot(
    age_range,
    strength_curve,
    linewidth=3,
    label="Polynomial Regression"
)

plt.xlabel("Age (days)")
plt.ylabel("Concrete Compressive Strength (MPa)")

plt.title(
    "Non-Linear Regression: Age vs Concrete Compressive Strength"
)

plt.legend()
plt.grid(True)

plt.show()
```

```python
#Model Evaluation

from sklearn.metrics import mean_squared_error, r2_score

# Calculate metrics
mse = mean_squared_error(
    strength,
    predicted_strength
)

r2 = r2_score(
    strength,
    predicted_strength
)

print("Mean Squared Error:", round(mse, 2))
print("R² Score:", round(r2, 4))
```

# **Result**

The Concrete Compressive Strength dataset was successfully used for Non-Linear Regression.

Polynomial Regression of degree 3 was implemented using the concrete age as the input variable and compressive strength as the output variable.

A graph was plotted to show the relationship between concrete age and predicted compressive strength.

The model was also evaluated using Mean Squared Error (MSE) and R² Score.
