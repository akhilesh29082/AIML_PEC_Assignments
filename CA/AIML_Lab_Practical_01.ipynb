# 5. PEAS Descriptors of Smart Hospital Vacuum Cleaner

PEAS stands for:

**P – Performance Measure**
**E – Environment**
**A – Actuators**
**S – Sensors**

The PEAS framework provides a systematic way of describing the task environment of an intelligent agent.

## PEAS Table

| PEAS Component          | Description                                                                                                                                                                                                               |
| ----------------------- | ------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------- |
| **Performance Measure** | Maximum cleanliness, minimum cleaning time, minimum battery consumption, safe navigation, successful obstacle avoidance, correct spill handling, prevention of entry into restricted areas, minimum unnecessary movements |
| **Environment**         | Patient rooms, corridors, waiting areas, ICU zones, hospital staff, patients, beds, wheelchairs, medical equipment, dirt, liquid spills, walls, doors, charging station                                                   |
| **Actuators**           | Wheels, movement motors, vacuum motor, suction controller, cleaning brushes, mop system, warning indicator, direction controller                                                                                          |
| **Sensors**             | Dirt sensor, obstacle sensor, proximity sensor, spill sensor, battery sensor, location sensor, restricted-zone sensor                                                                                                     |

## Performance Measure

The performance of the hospital vacuum cleaner is evaluated according to the following factors:

* Amount of dirt removed
* Number of areas cleaned
* Number of obstacles avoided
* Number of spills handled
* Number of restricted-area entries prevented
* Total battery consumed
* Number of unnecessary movements
* Successful return to charging station
* Overall performance score

The performance score increases when the vacuum cleaner successfully removes dirt, avoids obstacles, handles spills correctly, and prevents unsafe movement.

# 6. Conceptual Details

## Simple Reflex Agent

A **Simple Reflex Agent** is an intelligent agent that selects an action based only on the current condition of its environment.

The agent does not maintain a detailed history of previous states. Instead, it uses predefined **condition–action rules**.

The general working principle is:

**Sense Environment → Apply Condition–Action Rule → Perform Action**

In this practical, a Simple Reflex Agent controls a smart vacuum cleaner operating inside a hospital.

# 7. Hospital Environment

The hospital environment consists of the following areas:

* Patient Room
* Corridor
* Waiting Area
* ICU Zone

Each area contains the following properties:

* Dirt level
* Spill status
* Obstacle status
* Restricted-area status
* Cleaning status

The dirt level may be entered by the user, while other environmental conditions can be generated automatically during simulation.

# 8. Working of the Smart Hospital Vacuum Agent

The vacuum cleaner continuously senses its current environment and selects an action according to predefined conditions.

The main condition–action rules are:

| Condition                      | Action                            |
| ------------------------------ | --------------------------------- |
| Dirt level > 60%               | Use STRONG cleaning mode          |
| Dirt level between 31% and 60% | Use NORMAL cleaning mode          |
| Dirt level between 1% and 30%  | Use ECO cleaning mode             |
| Liquid spill detected          | Activate MOP mode                 |
| Obstacle detected              | Stop, reverse, and turn           |
| Person detected nearby         | Reduce speed and change direction |
| Restricted ICU zone detected   | Do not enter                      |
| Battery ≤ 20%                  | Return to charging station        |
| Area is already clean          | Move to another area              |
| All areas are clean            | Stop cleaning                     |

# 9. Cleaning Modes

The hospital vacuum cleaner uses different cleaning modes depending on the current dirt level.

## STRONG Mode

If the dirt level is greater than 60%, the cleaner uses high suction power.

**Action:**

* Strong suction
* High cleaning power
* Higher battery consumption

## NORMAL Mode

If the dirt level is between 31% and 60%, the cleaner uses normal suction power.

**Action:**

* Medium suction
* Moderate battery consumption

## ECO Mode

If the dirt level is between 1% and 30%, the cleaner uses low suction power.

**Action:**

* Low suction
* Minimum battery consumption

## MOP Mode

If a liquid spill is detected, the vacuum cleaner avoids normal suction and activates the mop system.

**Action:**

* Reduce movement speed
* Activate mop cleaning
* Clean the spill safely

# 10. Obstacle Handling

Hospitals contain many obstacles such as:

* Beds
* Wheelchairs
* Medical equipment
* Chairs
* Patients
* Doctors
* Nurses

If an obstacle is detected, the vacuum cleaner immediately:

1. Stops movement.
2. Reverses slightly.
3. Changes direction.
4. Continues cleaning.

If a person is detected nearby, the vacuum cleaner reduces its speed to ensure safe navigation.

# 11. Restricted ICU Area Handling

Certain hospital areas may have restricted access.

For example, an ICU section may temporarily restrict cleaning robots during medical procedures.

When the restricted-zone sensor detects a restricted ICU area:

1. The vacuum cleaner stops.
2. It does not enter the restricted zone.
3. It changes direction.
4. It selects another area for cleaning.

This improves safety and prevents interference with medical activities.

# 12. Spill Detection and Cleaning

Liquid spills are common in hospital environments.

When a spill is detected:

1. The spill sensor identifies the liquid.
2. Normal vacuum suction is disabled.
3. Mop mode is activated.
4. The cleaner safely cleans the liquid.
5. The area is marked as clean.

This prevents unsafe vacuuming of liquid material.

# 13. Battery Management

The vacuum cleaner begins operation with a battery level of **100%**.

Different actions consume different amounts of battery.

For example:

* ECO cleaning consumes less battery.
* NORMAL cleaning consumes moderate battery.
* STRONG cleaning consumes more battery.
* Movement consumes battery.
* Obstacle avoidance consumes additional battery.
* Spill cleaning consumes battery.

When the battery level reaches **20% or below**, the vacuum cleaner automatically returns to the charging station.

The charging station is assumed to be located near the hospital reception or maintenance room.

The vacuum cleaner recharges to 100% before continuing the cleaning process.

# 14. Area Selection

The vacuum cleaner compares the dirt level of different hospital areas and selects the area requiring the most cleaning.

For example:

**Patient Room = 20% dirt**
**Corridor = 75% dirt**
**Waiting Area = 45% dirt**
**ICU Zone = 30% dirt**

The vacuum cleaner selects:

**Corridor**

because the corridor currently has the highest dirt level.

However, if the corridor is marked as restricted, the vacuum cleaner selects the next available area.

# 15. Example Scenario

Consider the following hospital environment:

**Patient Room:** 25% dirt
**Corridor:** 70% dirt
**Waiting Area:** 45% dirt
**ICU Zone:** 80% dirt but restricted

The vacuum cleaner first identifies ICU Zone as the dirtiest area.

However, the ICU Zone is restricted.

Therefore, according to the condition–action rule:

**Restricted area detected → Do not enter**

The agent then selects the **Corridor**, which has 70% dirt.

Since 70% is greater than 60%, the agent uses:

**STRONG cleaning mode**

After cleaning the corridor, the vacuum cleaner again senses the current environment and selects the next appropriate area.

# 16. Performance Measurement

The overall working can be represented as:

**Sense Environment → Check Battery → Check Restricted Zone → Detect Obstacle → Detect Spill → Select Dirtiest Area → Select Cleaning Mode → Clean Area → Update Environment → Repeat**

The process continues until all non-restricted areas are clean.

# 20. Program


````python
import random
import time

# ---------------------------------------------
# SMART HOSPITAL VACUUM CLEANER
# SIMPLE REFLEX AGENT
# ---------------------------------------------

areas = {
    "Patient Room": {
        "dirt": 0,
        "spill": False,
        "obstacle": False,
        "restricted": False
    },
    "Corridor": {
        "dirt": 0,
        "spill": False,
        "obstacle": False,
        "restricted": False
    },
    "Waiting Area": {
        "dirt": 0,
        "spill": False,
        "obstacle": False,
        "restricted": False
    },
    "ICU Zone": {
        "dirt": 0,
        "spill": False,
        "obstacle": False,
        "restricted": True
    }
}

battery = 100
location = "Charging Station"

steps = 0
cleaning_operations = 0
dirt_removed = 0
obstacles_avoided = 0
spills_handled = 0
restricted_entries_prevented = 0
battery_used = 0
performance_score = 0


# Get dirt level
def get_dirt_level(area):
    while True:
        try:
            dirt = int(
                input(
                    f"Enter dirt level for {area} (0-100): "
                )
            )

            if 0 <= dirt <= 100:
                return dirt

            print("Enter a value between 0 and 100.")

        except ValueError:
            print("Enter a valid number.")


# Cleaning mode selection
def select_cleaning_mode(dirt):

    if dirt > 60:
        return "STRONG", 30, 7

    elif dirt > 30:
        return "NORMAL", 20, 4

    elif dirt > 0:
        return "ECO", 10, 2

    else:
        return "NONE", 0, 0


# Display hospital environment
def display_environment():

    print("\n" + "=" * 80)
    print("                     HOSPITAL ENVIRONMENT")
    print("=" * 80)

    for area, data in areas.items():

        spill = "YES" if data["spill"] else "NO"
        obstacle = "YES" if data["obstacle"] else "NO"
        restricted = "YES" if data["restricted"] else "NO"

        status = "CLEAN" if data["dirt"] == 0 else "DIRTY"

        print(
            f"{area:15} | "
            f"Dirt: {data['dirt']:3}% | "
            f"Spill: {spill:3} | "
            f"Obstacle: {obstacle:3} | "
            f"Restricted: {restricted:3} | "
            f"{status}"
        )

    print("=" * 80)


# Select dirtiest accessible area
def select_dirtiest_area():

    accessible_areas = {
        area: data["dirt"]
        for area, data in areas.items()
        if not data["restricted"] and data["dirt"] > 0
    }

    if not accessible_areas:
        return None

    return max(
        accessible_areas,
        key=accessible_areas.get
    )


# Check obstacle
def check_obstacle(area):

    global battery
    global battery_used
    global obstacles_avoided
    global performance_score

    if areas[area]["obstacle"]:

        print("\nObstacle detected!")
        print("Decision -> STOP -> REVERSE -> TURN")

        battery -= 2
        battery_used += 2

        obstacles_avoided += 1
        performance_score += 3

        areas[area]["obstacle"] = False

        return True

    return False


# Handle spill
def handle_spill(area):

    global battery
    global battery_used
    global spills_handled
    global performance_score

    if areas[area]["spill"]:

        print("\nLiquid spill detected!")
        print("Decision -> Activate MOP mode")

        battery -= 3
        battery_used += 3

        spills_handled += 1
        performance_score += 5

        areas[area]["spill"] = False

        print("Spill cleaned successfully.")

        return True

    return False


# Prevent restricted-area entry
def check_restricted_area(area):

    global restricted_entries_prevented
    global performance_score

    if areas[area]["restricted"]:

        print(
            f"\n{area} is currently RESTRICTED."
        )

        print(
            "Decision -> Do not enter. "
            "Select another area."
        )

        restricted_entries_prevented += 1
        performance_score += 5

        return True

    return False


# Clean selected area
def clean_area(area):

    global battery
    global battery_used
    global cleaning_operations
    global dirt_removed
    global performance_score

    current_dirt = areas[area]["dirt"]

    mode, cleaning_power, battery_cost = \
        select_cleaning_mode(current_dirt)

    if current_dirt == 0:
        return

    print(f"\nCleaning {area}")
    print(f"Cleaning Mode -> {mode}")

    old_dirt = current_dirt

    areas[area]["dirt"] = max(
        0,
        current_dirt - cleaning_power
    )

    removed = old_dirt - areas[area]["dirt"]

    battery -= battery_cost
    battery_used += battery_cost

    cleaning_operations += 1
    dirt_removed += removed

    performance_score += removed * 2

    print(f"Dirt Removed -> {removed}%")
    print(
        f"Remaining Dirt -> "
        f"{areas[area]['dirt']}%"
    )

    print(f"Battery -> {battery}%")


# Charging
def charge():

    global battery
    global location

    print("\nBATTERY LOW")
    print(f"Current Battery: {battery}%")
    print("Returning to Charging Station...")

    location = "Charging Station"

    print("Charging...")

    while battery < 100:

        battery += 10

        if battery > 100:
            battery = 100

        print(f"Battery: {battery}%")
        time.sleep(0.1)

    print("Battery fully charged.")


# Check if all accessible areas are clean
def all_areas_clean():

    for area, data in areas.items():

        if not data["restricted"] and data["dirt"] > 0:
            return False

    return True


# Simple Reflex Agent
def simple_reflex_agent():

    global steps
    global battery
    global location

    while not all_areas_clean():

        steps += 1

        print("\n" + "=" * 80)
        print(f"STEP {steps}")
        print("=" * 80)

        print(f"Current Location: {location}")
        print(f"Battery: {battery}%")

        # RULE 1 - Battery management
        if battery <= 20:

            charge()

            continue

        # RULE 2 - Select dirtiest accessible area
        target_area = select_dirtiest_area()

        if target_area is None:
            break

        print(
            f"Decision -> Select {target_area}"
        )

        location = target_area

        # RULE 3 - Restricted zone
        if check_restricted_area(target_area):

            continue

        # RULE 4 - Obstacle detection
        if check_obstacle(target_area):

            continue

        # RULE 5 - Spill detection
        handle_spill(target_area)

        # RULE 6 - Cleaning
        clean_area(target_area)

        time.sleep(0.2)


# ---------------------------------------------
# MAIN PROGRAM
# ---------------------------------------------

print("\n" + "=" * 80)
print("                 SMART HOSPITAL VACUUM CLEANER")
print("                     SIMPLE REFLEX AGENT")
print("=" * 80)

print("\nINITIAL SETUP")

for area in areas:

    areas[area]["dirt"] = get_dirt_level(area)


# Random obstacle and spill generation
for area in areas:

    areas[area]["obstacle"] = (
        random.random() < 0.25
    )

    areas[area]["spill"] = (
        random.random() < 0.20
    )


print("\nINITIAL ENVIRONMENT")

display_environment()

print("\nCharging Station: Hospital Maintenance Area")
print("Initial Battery: 100%")

print("\nSTARTING CLEANING PROCESS...")

simple_reflex_agent()


# ---------------------------------------------
# FINAL REPORT
# ---------------------------------------------

print("\n" + "=" * 80)
print("                         FINAL REPORT")
print("=" * 80)

print("\nCleaning process completed successfully.")

print("\nPERFORMANCE")
print("-" * 55)

print(f"Total Decision Steps            : {steps}")
print(f"Cleaning Operations             : {cleaning_operations}")
print(f"Total Dirt Removed              : {dirt_removed}%")
print(f"Obstacles Avoided               : {obstacles_avoided}")
print(f"Spills Handled                  : {spills_handled}")
print(
    f"Restricted Entries Prevented    : "
    f"{restricted_entries_prevented}"
)
print(f"Battery Used                    : {battery_used}%")
print(f"Final Battery Level             : {battery}%")
print(f"Performance Score               : {performance_score}")

print("\nFINAL ENVIRONMENT")

display_environment()

print("\nAll accessible hospital areas are clean.")
````

**Note:** The exact output can vary because obstacles and liquid spills are generated randomly during execution.
